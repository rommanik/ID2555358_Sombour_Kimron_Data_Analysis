# Part 1 — Presenting an analysis

## 1.1 The difference between a report and a talk

They are not the same artefact and they should not contain the same things.

| | Report (the notebook) | Talk (ten minutes) |
|---|---|---|
| Read at | the reader's pace | your pace |
| Detail | all of it | one level |
| Figures | as many as the argument needs | three, at most |
| Numbers per figure | as many as are relevant | one that matters |
| The audience can | re-read, scroll back, check | neither |
| Failure mode | missing detail | too much detail |

A talk is not a compressed report. It is a **different argument**, aimed at people who cannot
go back. Everything in it has to be understood the first time, which is why three figures is a
ceiling and not a target.

---

## 1.2 The ten-minute structure

| Block | Minutes | Content |
|---|---|---|
| **The question** | 1.5 | What you asked, and what would change depending on the answer |
| **The data** | 1.5 | Where it came from, what one row is, one thing that was wrong with it |
| **What you found** | 4 | Two or three findings, each with a number and a figure |
| **What you cannot say** | 1.5 | The limitation that matters most, named before anyone asks |
| **What should happen** | 1.5 | The recommendation, specific |

Two rules that do most of the work:

**Start with the answer.** Not the method. Your first slide after the question should contain
the finding. The audience then spends the rest of the talk understanding how you got there,
rather than waiting to find out where you are going.

**Name your own worst limitation before the audience does.** It costs thirty seconds and it
converts the most dangerous question in the room into something you have already answered.

---

## 1.3 Which figures survive a projector

A figure built for a report fails on a screen for reasons that have nothing to do with whether
it is a good figure.

| Report figure | Talk figure |
|---|---|
| Eight series, a legend | Two series, labelled directly on the lines |
| Title names the variables | Title **states the finding** |
| 10pt fonts | 14pt and up |
| Grid, ticks, annotations | Only what carries the point |
| The reader finds the message | The message is unmissable in two seconds |

The single most valuable change is the title. `"Wait time by session"` describes the axes.
`"Afternoons wait 7.0 minutes longer"` states the finding, and it keeps working after somebody
photographs the slide and looks at it a week later.

Section 2.2 builds both versions of the same figure.

---

## 1.4 The five questions you will be asked

Every presentation in this course gets some version of these. Prepare all five; they take two
minutes each.

1. **"How do you know it is not just chance?"** — Answer with the interval and the sample size,
   not with a p-value. Week 12 §2.6.
2. **"What else could explain that?"** — Name a confounder and say what you did about it. If you
   did nothing, say so. Week 13 §2.6.
3. **"How did you handle the missing data?"** — Say how much was missing, whether it was missing
   evenly across groups, and what you did. "I dropped them" is an acceptable answer; "I did not
   check" is not. Week 5.
4. **"Would this hold for other data?"** — Almost always no, or unknown. Say which one, and why.
5. **"What would you do with more time?"** — Have a specific answer ready. It is the easiest
   question in the room and a vague answer wastes it.

The honest answer is always available and it is always better received than a confident one
that does not survive a follow-up.

## 2.1 The café study as a ten-minute talk

Week 13's case study, reduced to what fits.

| Block | What is said | Figure |
|---|---|---|

persuasive for it: the audience has the same instinct the analyst had, and watching it fail is
what makes the correction land.

## 2.2 Figure triage

The same comparison, drawn twice.

```python
def build_wait_times(n=1728):
    """Waiting times at a district clinic, by session and department."""
    session = rng.choice(["morning", "afternoon"], size=n, p=[0.58, 0.42])
    department = rng.choice(["general", "maternal", "dental", "pharmacy"],
                            size=n, p=[0.46, 0.24, 0.14, 0.16])
    base = np.where(session == "afternoon", 33.8, 26.4)
    return pd.DataFrame({
        "session": session,
        "department": department,
        "wait_minutes": np.clip(rng.normal(base, 12.7, size=n), 2.0, None).round(1),
    })


waits = build_wait_times()
summary = waits.groupby(["department", "session"])["wait_minutes"].mean().unstack()
gap = (waits.loc[waits["session"] == "afternoon", "wait_minutes"].mean()
       - waits.loc[waits["session"] == "morning", "wait_minutes"].mean())
print(summary.round(2).to_string())
print(f"\noverall gap: {gap:.2f} minutes")
```

```python
fig, axes = plt.subplots(1, 2, figsize=(13.5, 4.6))

# ---------- the REPORT figure: complete, dense, read at the reader's pace ----------
departments = summary.index.tolist()
positions = np.arange(len(departments))
axes[0].bar(positions - 0.19, summary["morning"], width=0.38,
            color=PALETTE[0], label="morning")
axes[0].bar(positions + 0.19, summary["afternoon"], width=0.38,
            color=PALETTE[1], label="afternoon")
axes[0].set_xticks(positions, departments)
axes[0].set_title("Mean waiting time by department and session", loc="left")
axes[0].set_ylabel("minutes")
axes[0].set_xlabel("department")
axes[0].legend(fontsize=9, title="session")
axes[0].set_ylim(0)
for x, (morning, afternoon) in enumerate(zip(summary["morning"], summary["afternoon"])):
    axes[0].text(x - 0.19, morning + 0.6, f"{morning:.1f}", ha="center", fontsize=8)
    axes[0].text(x + 0.19, afternoon + 0.6, f"{afternoon:.1f}", ha="center", fontsize=8)

# ---------- the TALK figure: one message, readable from the back of the room ----------
means = waits.groupby("session")["wait_minutes"].mean()[["morning", "afternoon"]]
bars = axes[1].bar(["morning", "afternoon"], means, width=0.5,
                   color=[PALETTE[0], PALETTE[3]])
axes[1].set_title(f"Afternoons wait {gap:.1f} minutes longer", loc="left",
                  fontsize=16, fontweight="bold")
axes[1].set_ylabel("mean wait (minutes)", fontsize=13)
axes[1].tick_params(labelsize=13)
axes[1].set_ylim(0, means.max() * 1.25)
axes[1].grid(False)
for bar, value in zip(bars, means):
    axes[1].text(bar.get_x() + bar.get_width() / 2, value + 1.0, f"{value:.0f} min",
                 ha="center", fontsize=14, fontweight="bold")

plt.tight_layout()
plt.show()

print("Same data. The left one is correct and complete; the right one can be understood")
print("from the back of a room in two seconds, and its title survives a photograph.")
```

The right-hand figure **throws away** the department breakdown, the exact values, the legend and
the gridlines. That is not carelessness. Every one of those belongs in the report, where the
reader can study it; none of them survives ten seconds on a projector.

The changes that matter most, in order:

1. **The title states the finding**, not the axes.
2. **Fewer series.** Eight bars became two.
3. **Bigger type.** 13 to 16 point, not 8 to 10.
4. **Values on the marks**, so nobody reads the axis.

## 2.3 The questions, rehearsed

Five answers, prepared, for the café study. Each is under thirty seconds.

| Question | The answer |
|---|---|
| "How do you know it is not chance?" | "The interval runs from −10% to +9% on 276 branch-days. It is consistent with no effect and with a moderate one — which is why I am not claiming either." |
| "What else could explain it?" | "Three things did: prices rose two weeks before launch, the control branch was closed for a refit during the baseline, and the underlying trend is upward. The first two I corrected for; the third is what the control is for." |
| "How did you handle missing data?" | "There were no missing waiting times, but there were 15 impossible prices and 48 duplicated transaction ids. I removed both and reported the counts in section 4." |
| "Would this hold elsewhere?" | "Unknown. It is one chain of four branches in one city over fifteen months. The design would transfer; the number would not." |
| "What would you do with more time?" | "Stagger the rollout across ten branches and record launch dates and price changes in a table. That would measure the effect to about a point, instead of ten." |

Notice that every answer contains a number or a specific action. "I would look into that more" is
the answer that costs marks.

## 2.4 Check your own notebook before you submit

```python
REQUIRED_SECTIONS = ["1. Question", "2. The data", "3. Audit", "4. Cleaning",
                     "5. Exploration", "6. Analysis", "7. Findings",
                     "8. Limitations", "9. References"]


def notebook_stats(path) -> dict:
    """Read a .ipynb as JSON and report what is in it.

    A notebook is a JSON document: a dict with a "cells" list. No library is needed
    to inspect one, which is worth knowing the day a notebook will not open.
    """
    path = Path(path)
    if not path.exists():
        raise FileNotFoundError(f"no such notebook: {path}")
    document = json.loads(path.read_text(encoding="utf-8"))
    cells = document.get("cells", [])

    def source_of(cell):
        raw = cell.get("source", "")
        return raw if isinstance(raw, str) else "".join(raw)

    code_cells = [c for c in cells if c.get("cell_type") == "code"]
    headings = [line.strip().lstrip("# ").strip()
                for cell in cells if cell.get("cell_type") == "markdown"
                for line in source_of(cell).splitlines()
                if line.strip().startswith("## ")]

    return {
        "cells": len(cells),
        "code_cells": len(code_cells),
        "markdown_cells": len(cells) - len(code_cells),
        "unrun_code_cells": sum(1 for c in code_cells if c.get("execution_count") is None),
        "cells_with_output": sum(1 for c in code_cells if c.get("outputs")),
        "todo_cells": sum(1 for c in cells if "TODO" in source_of(c)),
        "headings": headings,
        "missing_sections": [s for s in REQUIRED_SECTIONS if s not in headings],
    }
```

```python
def write_demo_notebook(path, finished=True):
    """Write a small project notebook, either submission-ready or deliberately not."""
    cells = [{"cell_type": "markdown", "metadata": {},
              "source": "# Final Project — Demo\n"}]

    for index, section in enumerate(REQUIRED_SECTIONS):
        if not finished and section in ("7. Findings", "8. Limitations"):
            continue                                   # two sections never written
        cells.append({"cell_type": "markdown", "metadata": {},
                      "source": f"## {section}\n\nSome prose about {section.lower()}.\n"})
        source = f"print('{section}')\n"
        if not finished and section == "4. Cleaning":
            source = "# TODO: decide what to do about the duplicates\nprint('4. Cleaning')\n"
        cells.append({
            "cell_type": "code", "metadata": {}, "source": source,
            "execution_count": (index + 1) if finished else None,
            "outputs": ([{"output_type": "stream", "name": "stdout",
                          "text": f"{section}\n"}] if finished else []),
        })

    Path(path).write_text(json.dumps({
        "cells": cells,
        "metadata": {"kernelspec": {"display_name": "Python 3", "language": "python",
                                    "name": "python3"},
                     "language_info": {"name": "python"}},
        "nbformat": 4, "nbformat_minor": 5,
    }, indent=1), encoding="utf-8")
    return Path(path)


write_demo_notebook("good_project.ipynb", finished=True)
write_demo_notebook("weak_project.ipynb", finished=False)

for name in ["good_project.ipynb", "weak_project.ipynb"]:
    stats = notebook_stats(name)
    print(f"{name}")
    for key in ["cells", "code_cells", "markdown_cells", "unrun_code_cells",
                "cells_with_output", "todo_cells"]:
        print(f"    {key:<20} {stats[key]}")
    print(f"    missing_sections     {stats['missing_sections']}")
    print()
```

The weak notebook is not badly analysed. It may be excellent. What the check finds is that
**nobody can tell**: the cells were never run, there are no outputs to read, a `TODO` is still
in the cleaning section, and two of the nine sections do not exist — including the one worth the
most marks.

All four of those are fixable in ten minutes, and all four cost marks if they are still there
when the file is opened.

## 2.6 The traps, all in one cell

Every one of these caught somebody this semester. They run, so you can see the behaviour rather
than remember it.

```python
print("1. numpy and pandas disagree about ddof")
values = pd.Series([2.0, 4.0, 4.0, 4.0, 5.0, 5.0, 7.0, 9.0])
print(f"   values.std()            {values.std():.4f}   pandas defaults to ddof=1")
print(f"   values.to_numpy().std() {values.to_numpy().std():.4f}   numpy defaults to ddof=0")

print("\n2. bool is numeric to is_numeric_dtype and not to select_dtypes")
frame = pd.DataFrame({"n": [1, 2, 3], "flag": [True, False, True]})
print(f"   is_numeric_dtype(flag)             {pd.api.types.is_numeric_dtype(frame['flag'])}")
print(f"   select_dtypes(include='number')    {list(frame.select_dtypes(include='number'))}")

print("\n3. pandas aligns on the INDEX, not on position")
left = pd.Series([1, 2, 3], index=["a", "b", "c"])
right = pd.Series([10, 20, 30], index=["c", "b", "a"])
print(f"   left + right ->  {dict(left + right)}   (a+30, b+20, c+10)")

print("\n4. duplicated() marks the LATER copies; keep=False marks all of them")
ids = pd.Series(["x", "y", "x", "z", "y"])
print(f"   duplicated().sum()            {ids.duplicated().sum()}")
print(f"   duplicated(keep=False).sum()  {ids.duplicated(keep=False).sum()}")

print("\n5. Index.difference() sorts, so it breaks column-order assumptions")
columns = pd.Index(["zebra", "apple", "mango"])
print(f"   difference([])  ->  {list(columns.difference([]))}")

print("\n6. a merge on a non-unique key multiplies rows")
orders = pd.DataFrame({"customer": ["a", "b"], "amount": [10, 20]})
contacts = pd.DataFrame({"customer": ["a", "a", "b"], "phone": ["p1", "p2", "p3"]})
print(f"   2 orders + 3 contacts  ->  {len(orders.merge(contacts, on='customer'))} rows")
print("   validate='one_to_one' would have raised instead")
```

```python
print("7. the old frequency aliases were removed in pandas 3.0")
index = pd.date_range("2025-01-01", periods=90, freq="D")
series = pd.Series(np.arange(90, dtype=float), index=index)
for alias in ["ME", "M"]:
    try:
        series.resample(alias).mean()
        print(f"   freq={alias!r:>4}  works")
    except ValueError as error:
        print(f"   freq={alias!r:>4}  ValueError: {str(error)[:52]}...")

print("\n8. ax.get_title() returns only the CENTRE title")
figure, axis = plt.subplots(figsize=(4, 1.6))
axis.plot([0, 1], [0, 1], color=PALETTE[0])
axis.set_title("set with loc='left'", loc="left")
print(f"   get_title()              {axis.get_title()!r}   <- empty")
print(f"   get_title(loc='left')    {axis.get_title(loc='left')!r}")
plt.close(figure)

print("\n9. resampling destroys any cycle shorter than the period")
daily = pd.Series(np.where(index.dayofweek >= 5, 20.0, 100.0), index=index)
print(f"   daily variation   {daily.std(ddof=1):7.3f}")
print(f"   weekly means      {daily.resample('W').mean().std(ddof=1):7.3f}   <- the weekly "
      "cycle is gone")

print("\n10. chained assignment never lands under copy-on-write")
table = pd.DataFrame({"score": [1.0, np.nan, 3.0]})
try:
    table["score"][1] = 2.0
    print(f"   after chained assignment: {list(table['score'])}")
except Exception as error:
    print(f"   {type(error).__name__}: {str(error)[:60]}...")
print("   table.loc[1, 'score'] = 2.0  is the form that works")
```

Ten behaviours, all of them silent except the last two. Silent is what makes them expensive: the
code runs, the chart renders, and the number is wrong.

The habit that catches all ten is the same one from Week 13's audit: after any step that changes
a table, **look at the result** — its shape, its dtypes, and a few of its rows — rather than
assuming the step did what its name suggests.

# Part 3 — Guided workshop (in class)

Four tasks: one figure, one file check, and two pieces of your own presentation.

The helper below is used by task 1 and by the homework. Matplotlib keeps the left and right
titles in artists it does not expose publicly, so this reaches them defensively and falls back
to the centre title if a future version moves them.

```python
def title_artists(ax):
    """Every title artist on an Axes: centre, left and right."""
    candidates = [ax.title,
                  getattr(ax, "_left_title", None),
                  getattr(ax, "_right_title", None)]
    return [artist for artist in candidates if artist is not None]


def get_any_title(ax):
    """The Axes title, from whichever of the three positions it was set in.

    `ax.get_title()` returns only the centre title, so a title set with loc="left"
    reads as empty. This was Week 7's trap and it is still Week 15's.
    """
    for artist in title_artists(ax):
        if artist.get_text():
            return artist.get_text()
    return ""


demo_fig, demo_ax = plt.subplots(figsize=(4, 1.6))
demo_ax.plot([0, 1], [0, 1], color=PALETTE[0])
demo_ax.set_title("a left title", loc="left", fontsize=15)
print(f"ax.get_title()       {demo_ax.get_title()!r}")
print(f"get_any_title(ax)    {get_any_title(demo_ax)!r}")
print(f"its font size        {max(a.get_fontsize() for a in title_artists(demo_ax) if a.get_text())}")
plt.close(demo_fig)
```

```python
# Problem 1 — assertions. Run until this prints "all assertions passed".

def make_figure(title=None, title_size=14, xlabel="x", ylabel="y", n_series=2, panels=1):
    """Build a figure to order, for testing."""
    figure, axes = plt.subplots(1, panels, figsize=(4 * panels, 3), squeeze=False)
    for ax in axes.ravel():
        for index in range(n_series):
            ax.plot([1, 2, 3], [index, index + 1, index + 2])
        if xlabel is not None:
            ax.set_xlabel(xlabel)
        if ylabel is not None:
            ax.set_ylabel(ylabel)
    if title is not None:
        axes.ravel()[0].set_title(title, loc="left", fontsize=title_size)
    return figure


good = make_figure(title="Sales rose 12% in Q3")
assert check_figure(good) == [], f"A complete figure has no problems; got {check_figure(good)}"
plt.close(good)

# --- no title ---
untitled = make_figure(title=None)
found = check_figure(untitled)
assert any("title" in message for message in found), (
    f"A figure with no title must be reported; got {found}"
)
plt.close(untitled)

# --- title too small ---
small = make_figure(title="Sales rose 12% in Q3", title_size=9)
found = check_figure(small)
assert any("font" in message for message in found), (
    f"A 9pt title must be reported when min_title_size is 12; got {found}"
)
assert not any("title" in message and "font" not in message for message in found), (
    "A small title still counts as a title; do not also report it as missing"
)
plt.close(small)

# --- missing labels ---
no_x = make_figure(title="A title with 1 number", xlabel=None)
assert any("x-axis" in message for message in check_figure(no_x))
assert not any("y-axis" in message for message in check_figure(no_x))
plt.close(no_x)

no_y = make_figure(title="A title with 1 number", ylabel=None)
assert any("y-axis" in message for message in check_figure(no_y))
plt.close(no_y)

# --- too many series ---
busy = make_figure(title="A title with 1 number", n_series=6)
assert any("series" in message for message in check_figure(busy)), (
    "Six series with max_series=4 must be reported"
)
assert check_figure(busy, max_series=6) == [], (
    "With max_series=6 the same figure is fine"
)
plt.close(busy)

# --- per-axes checks fire once per axes ---
two_bad = make_figure(title="A title with 1 number", ylabel=None, panels=2)
found = check_figure(two_bad)
assert sum("y-axis" in message for message in found) == 2, (
    f"Two panels with no y label give two messages; got {found}"
)
plt.close(two_bad)

# --- the title check is about the whole figure, not each axes ---
one_title = make_figure(title="A title with 1 number", panels=2)
assert check_figure(one_title) == [], (
    "One title on a two-panel figure is enough; the check is per FIGURE"
)
plt.close(one_title)

# --- sorted output ---
messy = make_figure(title=None, xlabel=None, ylabel=None, n_series=9)
found = check_figure(messy)
assert found == sorted(found), f"The list must be sorted; got {found}"
assert len(found) >= 4, f"Four things are wrong with that figure; found {len(found)}"
plt.close(messy)

# --- bar containers count as series too ---
bar_fig, bar_ax = plt.subplots()
bar_ax.bar([0, 1], [1, 2], width=0.4)
bar_ax.set_title("Bars beat lines by 2", fontsize=14)
bar_ax.set_xlabel("group")
bar_ax.set_ylabel("value")
assert check_figure(bar_fig) == [], f"A labelled bar chart is fine; got {check_figure(bar_fig)}"

bar_ax.bar([0.5, 1.5], [3, 4], width=0.4)          # a second bar container
assert any("series" in message for message in check_figure(bar_fig, max_series=1)), (
    "Two bar containers with max_series=1 must be reported"
)
plt.close(bar_fig)

# --- errors ---
for bad in ["not a figure", None, 42]:
    try:
        check_figure(bad)
    except TypeError:
        pass
    else:
        raise AssertionError(f"{type(bad).__name__} input must raise TypeError")

spare = make_figure(title="A title with 1 number")
for bad_max in [0, -1]:
    try:
        check_figure(spare, max_series=bad_max)
    except ValueError:
        pass
    else:
        raise AssertionError(f"max_series={bad_max} must raise ValueError")
for bad_size in [0, -3]:
    try:
        check_figure(spare, min_title_size=bad_size)
    except ValueError:
        pass
    else:
        raise AssertionError(f"min_title_size={bad_size} must raise ValueError")
plt.close(spare)

print("Problem 1: all assertions passed.")
```

```python
# Problem 2 — assertions. Run until this prints "all assertions passed".

def write_notebook(path, cells):
    """Minimal notebook writer for the tests."""
    Path(path).write_text(json.dumps({
        "cells": cells, "metadata": {}, "nbformat": 4, "nbformat_minor": 5,
    }, indent=1), encoding="utf-8")
    return path


def markdown(text):
    return {"cell_type": "markdown", "metadata": {}, "source": text}


def code_cell(source, run=True, output=True):
    return {"cell_type": "code", "metadata": {}, "source": source,
            "execution_count": 1 if run else None,
            "outputs": ([{"output_type": "stream", "name": "stdout", "text": "ok\n"}]
                        if output else [])}


SECTIONS = ["A", "B", "C"]
complete = [markdown("## A\n"), code_cell("print(1)\n"),
            markdown("## B\n"), code_cell("print(2)\n"),
            markdown("## C\n"), code_cell("print(3)\n")]

write_notebook("t_complete.ipynb", complete)
assert check_submission("t_complete.ipynb", SECTIONS) == [], (
    f"A complete notebook has no problems; got {check_submission('t_complete.ipynb', SECTIONS)}"
)

# --- a missing section ---
write_notebook("t_missing.ipynb", complete[:4])
found = check_submission("t_missing.ipynb", SECTIONS)
assert any("missing section" in message for message in found), found
assert sum("missing section" in message for message in found) == 1, (
    f"One section is missing, so one message; got {found}"
)

# --- out of order ---
write_notebook("t_order.ipynb", [complete[2], complete[3], complete[0], complete[1],
                                 complete[4], complete[5]])
found = check_submission("t_order.ipynb", SECTIONS)
assert any("order" in message for message in found), found
assert not any("missing section" in message for message in found), (
    f"Nothing is missing, only out of order; got {found}"
)

# --- a leftover TODO ---
write_notebook("t_todo.ipynb", complete[:4] + [markdown("## C\n"),
                                               code_cell("# TODO: finish this\n")])
assert any("TODO" in message for message in check_submission("t_todo.ipynb", SECTIONS))

# --- an unrun cell ---
write_notebook("t_unrun.ipynb", complete[:4] + [markdown("## C\n"),
                                                code_cell("print(3)\n", run=False)])
assert any("not been run" in message
           for message in check_submission("t_unrun.ipynb", SECTIONS))

# --- no outputs at all ---
write_notebook("t_nooutput.ipynb",
               [markdown("## A\n"), code_cell("print(1)\n", output=False),
                markdown("## B\n"), code_cell("print(2)\n", output=False),
                markdown("## C\n"), code_cell("print(3)\n", output=False)])
assert any("output" in message
           for message in check_submission("t_nooutput.ipynb", SECTIONS))

# --- source given as a list of lines, which is how Jupyter usually writes it ---
write_notebook("t_list.ipynb", [
    {"cell_type": "markdown", "metadata": {}, "source": ["## A\n"]},
    {"cell_type": "code", "metadata": {}, "source": ["print(", "1)\n"],
     "execution_count": 1,
     "outputs": [{"output_type": "stream", "name": "stdout", "text": "1\n"}]},
    markdown("## B\n"), code_cell("print(2)\n"),
    markdown("## C\n"), code_cell("print(3)\n"),
])
assert check_submission("t_list.ipynb", SECTIONS) == [], (
    "A 'source' that is a list of lines must be handled"
)

# --- sorted ---
write_notebook("t_messy.ipynb", [markdown("## B\n"),
                                 code_cell("# TODO\n", run=False, output=False),
                                 markdown("## A\n"), code_cell("print(2)\n", run=False,
                                                               output=False)])
found = check_submission("t_messy.ipynb", SECTIONS)
assert found == sorted(found), f"The list must be sorted; got {found}"
assert len(found) >= 4, f"Four things are wrong with that notebook; found {found}"

# --- errors ---
try:
    check_submission("does_not_exist.ipynb", SECTIONS)
except FileNotFoundError:
    pass
else:
    raise AssertionError("A missing file must raise FileNotFoundError")

try:
    check_submission("t_complete.ipynb", "A")
except TypeError:
    pass
else:
    raise AssertionError("A string for required_sections must raise TypeError")

Path("t_broken.ipynb").write_text("{not json at all", encoding="utf-8")
try:
    check_submission("t_broken.ipynb", SECTIONS)
except ValueError:
    pass
else:
    raise AssertionError("A file that is not JSON must raise ValueError")

Path("t_nocells.ipynb").write_text(json.dumps({"metadata": {}}), encoding="utf-8")
try:
    check_submission("t_nocells.ipynb", SECTIONS)
except ValueError:
    pass
else:
    raise AssertionError("JSON with no 'cells' key must raise ValueError")

# --- and it agrees with section 2.4 on the two demo notebooks ---
assert check_submission("good_project.ipynb", REQUIRED_SECTIONS) == []
weak = check_submission("weak_project.ipynb", REQUIRED_SECTIONS)
assert sum("missing section" in message for message in weak) == 2, (
    f"The weak notebook is missing Findings and Limitations; got {weak}"
)

print("Problem 2: all assertions passed.")
```
